```python
import nbformat
import time

start_time = time.time()
notebook_files = ['DMW_GA2_Generate_Synthetic_Data_(Cleaned).ipynb', 'DMW_Decision_Tree (Amni).ipynb', 'DMW_Logistic_Regression (Aliya).ipynb', 'DMW_Naive_Bayes_Model (Hani).ipynb', 'DMW_MLP (Iskandar).ipynb', 'DMW_Random_Forest_Model (Xayne).ipynb']
final_notebook = nbformat.v4.new_notebook()
all_cells = []

for notebook_file in notebook_files:
    read_start = time.time()
    with open(notebook_file, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
        all_cells.extend(nb.cells)
    print(f"Read {notebook_file} in {time.time() - read_start:.2f} seconds")

final_notebook.cells = all_cells
write_start = time.time()

with open('merged_notebook.ipynb', 'w', encoding='utf-8') as f:
    nbformat.write(final_notebook, f)

print(f"Wrote merged_notebook.ipynb in {time.time() - write_start:.2f} seconds")
print(f"Total time: {time.time() - start_time:.2f} seconds")
```
